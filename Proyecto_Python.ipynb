Hay duplicados dentro de categoria, eliminanos los duplicados de tag

```python
def remove_duplicate_tags(tags_str):
    # Si el valor es nulo o no es texto, lo dejamos igual
    if pd.isna(tags_str) or not isinstance(tags_str, str):
        return tags_str

    # 1. Separar por coma y quitar espacios en blanco alrededor de cada tag
    tags = [t.strip() for t in tags_str.split(',')]

    # 2. Eliminar duplicados manteniendo el orden
    unique_tags = list(dict.fromkeys(tags))

    # 3. Volver a unir los tags limpios
    return ", ".join(unique_tags)

# Aplicar la función a la columna 'tags'
df['tags'] = df['tags'].apply(remove_duplicate_tags)
```

En una variable hay más de un elemento, por eso creamos nueva columna para genero, donde hay solo 3 opciones, hombre, mujer, o varios para mas de una persona




```python
def clasificar_genero(valor):
    # Si es nulo o no es texto
    if pd.isna(valor):
        return 'None'

    # Limpiar y pasar a minúsculas
    texto = str(valor).lower().strip()

    # Separar la lista por comas
    generos = [g.strip() for g in texto.split(',') if g.strip()]

    # Si hay más de un solicitante listado
    if len(generos) > 1:
        return 'several'

    # Si hay exactamente uno:
    if len(generos) == 1:
        g = generos[0]
        if g == 'female':
            return 'female'
        elif g == 'male':
            return 'male'

    return 'other'

# Aplicar la función a la columna
col_genero = 'borrower_genders'
df['gender_clean'] = df[col_genero].apply(clasificar_genero)

# Verificar los resultados
print(df['gender_clean'].value_counts(dropna=False))
```

# OUTLIERS para las variables numericas

Para detectar outliers usamos graficos de boxplot y histograma

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Seleccionar  todas las columnas numéricas

num_cols = ['funded_amount', 'loan_amount', 'term_in_months', 'lender_count']

# 2. Configuración estética
sns.set_theme(style="whitegrid")

# 3. Iterar y graficar histograma + boxplot para cada variable
for col in num_cols:
    # Descartar nulos temporalmente para la gráfica
    data_col = df[col].dropna()

    fig, axes = plt.subplots(1, 2, figsize=(14, 4))

    # Histograma con KDE (curva de densidad)
    sns.histplot(data_col, kde=True, ax=axes[0], color='skyblue', bins=40)
    axes[0].set_title(f'Histograma de {col}', fontsize=12, fontweight='bold')
    axes[0].set_xlabel(col)
    axes[0].set_ylabel('Frecuencia')

    # Boxplot
    sns.boxplot(x=data_col, ax=axes[1], color='salmon', flierprops={'marker': 'o', 'markersize': 3, 'alpha': 0.5})
    axes[1].set_title(f'Boxplot de {col}', fontsize=12, fontweight='bold')
    axes[1].set_xlabel(col)

    plt.tight_layout()
    plt.show()
```

A partir de los graficos se ve que hay las variables titnen la distribución no de la Campana de Gaus, con outliers.

Detectamos matemáticamente los límites con IQR (Rango Intercuartílico) para saber exactamente qué valores son considerados outliers numéricamente según la regla del boxplot

```python
for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lim_inferior = q1 - 1.5 * iqr
    lim_superior = q3 + 1.5 * iqr

    outliers = df[(df[col] < lim_inferior) | (df[col] > lim_superior)][col]
    porcentaje = (len(outliers) / len(df)) * 100

    print(f"--- {col} ---")
    print(f"Límite superior: {lim_superior:,.2f}")
    print(f"Total outliers: {len(outliers):,} ({porcentaje:.2f}% del dataset)\n")
```

No es posible eliminar los outliers, ya que podemos perder data importante, en vez de eliminar creamos nuevas variables con valores logaritmicos para unificar la distibución.

Compruebo otro metodo para trabajar con outliers: IQR para variable lander_count. IQR se basa en la mediana y los cuartiles, por lo que no se ve afectado por valores extremos o datos muy sesgados.


```python
col = 'lender_count'

# 1. Cuartiles y Rango Intercuartílico (IQR)
q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

# 2. Límites de Tukey
lim_inferior = q1 - 1.5 * iqr
lim_superior = q3 + 1.5 * iqr

# 3. Filtrar outliers
outliers = df[(df[col] < lim_inferior) | (df[col] > lim_superior)][col]
porcentaje = (len(outliers) / len(df)) * 100

# 4. Resultados
print(f"--- {col} ---")
print(f"Q1 (25%): {q1:,.2f} | Q3 (75%): {q3:,.2f} | IQR: {iqr:,.2f}")
print(f"Límite inferior: {lim_inferior:,.2f}")
print(f"Límite superior: {lim_superior:,.2f}")
print(f"Total outliers: {len(outliers):,} ({porcentaje:.2f}% del dataset)")
```
